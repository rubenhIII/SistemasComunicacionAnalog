```python
from scipy.io import wavfile
```

```python
# Leo el archivo de audio
# Me regresa dos valores: frecuencia de muestreo y dos arrays de muestras (Estereo)
nombre_archivo = "instrumentos.wav"
frec_muestreo, datos = wavfile.read(nombre_archivo)
```

```python
# Imprimo la frec muestreo
print("Frecuencia de muestreo", frec_muestreo)
```

```python
# Imprimo el tipo y forma de los datos
print(type(datos))
print(datos.shape)
```

```python
# Junto en un solo canal los dos estereos
señal = datos[:,0] + datos[:,1]
len(señal)
```

```python
señal = señal[0:100000]
```

```python
import matplotlib.pyplot as plt
import numpy as np

eje_x = np.array(range(len(señal))) / frec_muestreo

plt.figure(figsize=(15,5))

plt.xlabel("Tiempo en segundos")
plt.ylabel("Amplitud")

plt.plot(eje_x, señal)
plt.show()
```
