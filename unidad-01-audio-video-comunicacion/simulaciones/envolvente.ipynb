```python
from scipy.io import wavfile
```

```python
nombre_archivo = "instrumentos.wav"
frec_muestreo, datos = wavfile.read(nombre_archivo)
```

```python
print("Frecuencia de muestreo", frec_muestreo)
```

```python
print(type(datos))
print(datos.shape)
```

```python
señal = datos[:,0] + datos[:,1]
len(señal)
```

```python
import matplotlib.pyplot as plt
import numpy as np

eje_x = np.array(range(len(señal))) / frec_muestreo

plt.figure(figsize=(15,5))

plt.xlabel("Tiempo en segundos")
plt.ylabel("Amplitud")

plt.plot(eje_x, señal)
plt.show()
```
