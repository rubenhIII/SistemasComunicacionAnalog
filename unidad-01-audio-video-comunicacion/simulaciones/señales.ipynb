```python
import sounddevice as sd
import numpy as np
import matplotlib.pyplot as plt


def generar_senal(frecuencia=440, fs=44100, duracion=1):
    amplitud = 1
    tiempo = np.linspace(0, duracion, int(fs * duracion), endpoint=False)
    senal = amplitud * np.sin(2 * np.pi * frecuencia * tiempo)
    return senal, tiempo

def mostrar_senal(senal, tiempo):
    plt.figure(figsize=(12, 4))
    plt.plot(tiempo, senal)
    plt.title('Grafica de Señal')
    plt.xlabel('Tiempo (s)')
    plt.ylim(2*np.min(senal), 2*np.max(senal))
    plt.ylabel('Amplitud')
    plt.grid(True)
    plt.show()

def reproducir(senal, fs=44100):
    sd.play(senal, fs)
    sd.wait()
```

```python
s, t = generar_senal(frecuencia=220)
mostrar_senal(s[0:1000], t[0:1000])
```

```python
reproducir(s)
```
