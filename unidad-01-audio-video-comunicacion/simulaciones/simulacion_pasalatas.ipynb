```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = np.linspace(0, 15000, 15001)
w = 2 * np.pi * f
```

```python
R = 1E3
C = 1E-6
```

```python
transformada = (1j * w) / (1 + ((1 / (R * C)) * 1j * w))
```

```python
m = np.abs(transformada)
p = np.angle(transformada)
```

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 8))
axes[0].plot(f, m)
axes[0].set_xscale('log')
axes[0].set_yscale('log')

axes[1].plot(f, p)
axes[1].set_xscale('log')
axes[1].set_yscale('log')
```
